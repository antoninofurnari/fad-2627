An analysis that nobody understands has not been completed. This lecture is about the
last step: turning a set of findings into something an audience can follow, act on, and
challenge.

It sits deliberately at the end of Block 1, right after the
[causal chapter](11_causal_analysis.qmd). By this point you can establish what the data
supports; the risk from here on is that a confident chart says more than the analysis
does. Most misleading data graphics are not produced by liars — they are produced by
people who wanted to make a point clearly.

## What the lecture covers

The material follows the structure of Knaflic's book:

1. **Understand the context.** Who is the audience, what do they already know, what
   decision are they going to make, and what exactly do you need them to do? A chart for
   yourself and a chart for a stakeholder are different objects.
2. **Choose an effective visual.** Match the display to the question — the comparison
   you want the reader to make should be the easiest one to make on the page. Chapter 2
   of these notes covers the mechanics; this is about the choice.
3. **Eliminate clutter.** Every element that is not carrying information is competing
   with the ones that are: gridlines, borders, legends that could be direct labels,
   decoration, the third dimension of a bar chart.
4. **Focus attention.** Colour, size and position are the tools that say *look here*.
   They work only if they are used sparingly: if everything is highlighted, nothing is.
5. **Think like a designer.** Affordance, accessibility and aesthetics — make the chart
   look like what it does, make it readable to a colour-blind reader and on a projector,
   and accept that a tidy chart is trusted more than an untidy one.
6. **Tell a story.** Narrative structure — situation, complication, resolution — applied
   to an analysis, so the audience remembers the point rather than the plot.

## The part that belongs to this course

Two obligations run alongside the craft, and they are what we will assess.

**A chart must not claim more than the analysis supports.** The most common failure in
project presentations is a slide showing a clean association with a title written in
causal language. If chapter 11 could not close the backdoor paths, the title cannot
either. This is why the project's slide deck must include a "What could confound this?"
slide: it forces the question onto the same page as the claim.

**Uncertainty is part of the result, not a caveat to be dropped for clarity.** A point
estimate shown without its interval is a cleaner chart and a worse one. Error bars, a
confidence band, or a sentence stating the interval — one of the three has to be there.
